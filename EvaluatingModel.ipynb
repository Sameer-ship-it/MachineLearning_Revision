```python
from sklearn.metrics import accuracy_score , precision_score , recall_score , f1_score

#Actual data(what happend actually)

y_true = [ 1 , 0 , 1 , 1 , 0 , 0]

#predicted data(what machine predicted)

y_pred = [ 1 , 0 , 1 , 1 , 1 , 0]

print ("Accuracy : ", accuracy_score(y_true , y_pred))
print("precision :" , precision_score(y_true , y_pred))
print("Recall : " , recall_score(y_true , y_pred))
print("F1 Score : " , f1_score(y_true , y_pred))
```

```python
'''Confusion Metrics'''

from sklearn.metrics import confusion_matrix

y_true = [ 1 , 0 , 1 , 1 , 0 , 0]
y_pred = [ 1 , 1 , 0, 1 , 0 , 0]


print("Confusion Matrix : ", confusion_matrix(y_true, y_pred))



```

```python
'''MAE , MSE , RMSE '''
from sklearn.metrics import mean_absolute_error , mean_squared_error , root_mean_squared_error
import numpy as np

# Real Scores

real_scores = [ 90 , 60 , 80 , 100]

# Predicted Score

pred_scores = [ 85 , 70 , 70 , 95 ]

print("MAE :", mean_absolute_error(real_scores, pred_scores))
print("MSE :", mean_squared_error(real_scores, pred_scores))
print("RMSE :", root_mean_squared_error(real_scores, pred_scores))






```

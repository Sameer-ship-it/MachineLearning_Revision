```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

data = {
    'Customer' : ['Riya','Sameer','faizaan','Neha','Sneha'],
    'Age': [ 20,21,21,13,15],
    'Spending':[100,5000,1200,400,300]
}
df = pd.DataFrame(data)

X = df[['Age' , 'Spending']]

model = KMeans(n_clusters = 2, random_state = 42, n_init = 10)
df['Group'] = model.fit_predict(X)

plt.figure(figsize = (6 , 5))
for group in df['Group'].unique():
    plt.scatter(
        df[df['Group'] == group]['Age'],
        df[df['Group'] == group]['Spending'],
        label=f'Group {group}'
    )
plt.xlabel('Age')
plt.ylabel('Spending')
plt.legend()
plt.show()

```

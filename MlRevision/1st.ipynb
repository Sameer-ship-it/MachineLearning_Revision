```python
import pandas as pd
df = pd.read_csv(r"C:\Users\shima\Downloads\Students_Performance_Dataset.csv")
```

```python
display(df.head())
display(df.tail())
print(df.isnull().sum())
print("\n")
df.info()
print("\n")
print(df.describe())
```

```python
df.dropna(inplace = True)
print(df)
print(df.isnull().sum)
```

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler
scaler = StandardScaler()
le = LabelEncoder()
df['Internet'] = le.fit_transform(df['Internet_Access_at_Home'])
display(df.head())


```

```python
X = df[['Study_Hours_per_Week','Internet']]
y = df[['Total_Score']]

X_scaled = scaler.fit_transform(X)
y_scaled = scaler.fit_transform(y)
print(X_scaled)
print(y_scaled)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X_scaled, y_scaled , test_size=0.2 , random_state=42) 
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_absolute_percentage_error,mean_squared_error,r2_score

model = LinearRegression()
model.fit (X_train,y_train)
y_pred = model.predict(X_test)

mape = mean_absolute_percentage_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print('Mean Absolute percentage', mape)
print("Mean Squared Error:", mse)
print("Mean Absolute Error:", mae)
print("R2 Score:", r2)

'''ye ek aise hi random trainin hai model ki jisme 
output ki accuracy bht kharab hai, ye without seriousness
revision h jiska aim bs revise krna tha
'''






```

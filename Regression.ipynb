```python
from sklearn.linear_model import LinearRegression
```

```python
X = [[1],[2],[3],[4],[5]]
y = [40 , 50 , 65 , 75 , 90]

model = LinearRegression() #LinearRegression for contnous value 

model.fit(X, y)

hours = float(input("Enter how many hours you sudied = "))
predicted_marks = model.predict([[hours]])

print(f'based on your hours {hours} you may score around {predicted_marks}')
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
#Logistic Regression For Binary Output

X = [[1],[2],[3],[4],[5]] #Studied hour input
y = [ 0 , 0 , 1 , 1 , 1] #result 0 for fail , 1 for p

model = LogisticRegression()

model.fit(X , y)

hours = float(input("Enter how many hours you sudied = "))
result = model.predict([[hours]])[0]

if result==1:
    print(f'based on your hours {hours} you are likly to pass')
else:
    print('You are fail')
```

```python
from sklearn.neighbors import KNeighborsClassifier
x =[
    [180,7],
    [200,7.5],
    [250,8],
    [300,8.5],
    [330,9],
    [360,9.5]
]
# 0 FOR Apple , 1 for Orange
y = [1,1,1,0,0,0]
model = KNeighborsClassifier(n_neighbors=3)
model.fit(x, y)

weight = float(input("Enter the weight in grams = "))
size = float(input("Enter the size in centimeters = "))

prediction = model.predict([[weight , size ]])[0]

if prediction == 0:
    print("Likely an Apple")
else:
    print("Likly an Orange")




```

```python
'''DECISION TREE'''
from sklearn.tree import DecisionTreeClassifier
X = [
    [7,2],
    [8,3],
    [9,8],
    [10,9]
]

y = [0,0,1,1] # 0 = Apple , 1 == orange

model = DecisionTreeClassifier()

model.fit(X , y)


Shade = float(input("Enter the color Shade(1-10) = "))
size = float(input("Enter the size in centimeters = "))

prediction = model.predict([[Shade , size ]])[0]

if prediction == 0:
    print("Likely an Apple")
else:
    print("Likly an Orange")



```
